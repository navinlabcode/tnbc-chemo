```python
meta_fields = [
    'paper_patient_id', 
    'cell_type', 
    'cell_state',
    'pCR_status', 
    'cell_type_ontology_term_id',
    'assay_ontology_term_id', 
    'development_stage_ontology_term_id', 
    'self_reported_ethnicity_ontology_term_id', 
    'sex_ontology_term_id', 
    'tissue_ontology_term_id', 
    'disease_ontology_term_id', 
    'suspension_type', 
    'tissue_type',
    'donor_id'
]

for field in meta_fields:
    if field in adata.obs.columns:
        print(f"Plotting: {field}")
        sc.pl.umap(
            adata,
            color=field,
            # legend_loc='on data',
            frameon=False,
            title=field,
            show=True
        )
    else:
        print(f"⚠️ Skipped {field} — not found in adata.obs.")
```
